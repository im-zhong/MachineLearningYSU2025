```python
import torch
from torch import nn, Tensor
```

```python


batch_size = 4
out_features = 2

input = torch.randn(size=(batch_size, out_features))
print(input)
```

```python
torch_relu = nn.ReLU()
output = torch_relu(input)
print(output)
```

```python
class MyReLU(nn.Module):
    def __init__(self) -> None:
        super().__init__()
        pass

    def forward(self, x: Tensor) -> Tensor:
        # TODO:
        zero = torch.zeros_like(x)
        # element-wise
        return torch.max(x, zero)
```

```python
# use myrelu
my_relu = MyReLU()
my_output = my_relu(input)
print(my_output)
```

```python
x1 = 0.1
x2 = 0.100000001

print(x1 == x2)
print(x1 - x2 <= 1e6)
```

```python
assert torch.all(torch.isclose(output, my_output))
```
