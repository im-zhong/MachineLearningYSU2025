```python
import torch
from torch import nn
from torch import Tensor
```

```python
def linear(X: Tensor, W: Tensor, B: Tensor) -> Tensor:
    output = X @ W + B
    return output
```

```python
class MyLinear(nn.Module):
    def __init__(self, in_features: int, out_features) -> None:
        super().__init__()

        # define a W matirx and B matrix
        self.W = nn.Parameter(torch.randn(size=(in_features, out_features)))
        # self.B = # TODO
        # self.B = nn.Parameter(torch.randn(size=(out_features,)))
        self.B = torch.randn(size=(out_features,))

    def forward(self, input: Tensor) -> Tensor:
        output = input @ self.W + self.B
        return output
```

```python
in_features = 8
out_features = 2
batch_size = 4

torch_linear = torch.nn.Linear(in_features=in_features, out_features=out_features)

input = torch.randn(size=(batch_size, in_features))

torch_output = torch_linear(input)
print(torch_output.shape)
```

```python
my_linear = MyLinear(in_features=in_features, out_features=out_features)
my_output = my_linear(input)
print(my_output.shape)
```

```python
print(my_linear.W.requires_grad, my_linear.B.requires_grad)

my_linear.W.requires_grad = False
```
