```python
import torch
from torch import nn, Tensor
```

```python
torch_mse = nn.MSELoss()

batch_size = 4
out_feature = 2

input = torch.randn(size=(batch_size, out_feature))
target = torch.randn(size=(batch_size, out_feature))

torch_loss = torch_mse(input, target)
print(torch_loss)
```

```python
class MyMSELoss(nn.Module):
    def __init__(self) -> None:
        super().__init__()
        pass


    def forward(self, input: Tensor, target: Tensor) -> Tensor:
        # TODO:
        return torch.mean((input - target) ** 2)

```

```python
my_mse = MyMSELoss()
my_loss = my_mse(input, target)
print(my_loss)
```

```python
assert torch.isclose(my_loss, torch_loss)
```
